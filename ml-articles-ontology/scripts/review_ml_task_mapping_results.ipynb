# Review ML task mapping results

```python
# Imports

import pandas as pd
import numpy as np  
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load data

results_path = "../results/ml_task_mapping/"
extended_results_path = "../results/extended_ml_task_mapping/"

task_matches_best = pd.read_csv(extended_results_path + "task_matches_best.csv")
task_matches_topk = pd.read_csv(extended_results_path + "task_matches_topk.csv")
task_matches_no_match = pd.read_csv(extended_results_path + "task_matches_no_match.csv")
```

```python
# Check task distribution in best matches
plt.figure(figsize=(12,6))
sns.countplot(data=task_matches_best, y="taskLabel", order=task_matches_best["taskLabel"].value_counts().index)
plt.title("Distribution of ML Tasks in Best Matches")
plt.xlabel("Count")
plt.ylabel("ML Task")
plt.show()
```
